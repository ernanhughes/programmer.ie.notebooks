```python
from evidence_pipeline import sparse_dot, maxsim
import math
from semantic_search import normalise, dot

q = {"cache": 1.0, "purge": 0.5}
d = {"cache": 0.7, "purge": 0.2}
score = sparse_dot(q, d)
print("sparse_dot:", score)
assert abs(score - 0.8) < 1e-12
assert sparse_dot({"absent": 1.0}, {}) == 0.0

try:
    sparse_dot({"bad": -1.0}, {})
    raised = False
except ValueError:
    raised = True
assert raised
```

```python
query = [[1, 0], [0, 1]]
separate = [[1, 0], [0, 1]]
collapsed = [[1, 1]]
print("MaxSim separate:", maxsim(query, separate))
print("MaxSim collapsed:", maxsim(query, collapsed))
assert maxsim(query, separate) == 2
assert abs(maxsim(query, collapsed) - math.sqrt(2)) < 1e-12

pq = normalise([0.5, 0.5])
ps = normalise([0.5, 0.5])
pc = normalise([1.0, 1.0])
print("pooled cosine separate:", dot(pq, ps))
print("pooled cosine collapsed:", dot(pq, pc))
assert abs(dot(pq, ps) - 1.0) < 1e-12
assert abs(dot(pq, pc) - 1.0) < 1e-12

try:
    maxsim(query, [])
    raised = False
except ValueError:
    raised = True
assert raised
```

No SPLADE or ColBERT model was evaluated in this notebook. The values above are hand-authored teaching vectors; accuracy, index size, latency and memory remain not measured.
