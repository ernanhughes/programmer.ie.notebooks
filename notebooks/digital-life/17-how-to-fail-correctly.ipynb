one asks "did we correctly *say* what we found?" -- did an invalid run get miscounted as a negative result,
did a bounded-near-zero result get called FAILED, did a stronger control silently erase a real earlier

**What this notebook does that the legacy pure-audit pattern did not:** it does not just narrate the ledger

`16-causal-modularity-vs-matched-null.ipynb`'s own executed outputs elsewhere in this notebook suite --
not by re-citing the archived `ch28` verdict, but by checking what this reconstruction's *own* Chapter 16
notebook actually computed. That is the chapter's own in-universe idiom (Ch.14-C01's "ASSERTION/IDENTITY"
framing: some things are correctness facts to be checked, not findings to be argued) applied to this
reconstruction project itself.

## Question

Across the registered Chapter 14-16 (original Ch.26-28) evidence transitions -- an invalidated
implementation, a direction-supported-but-magnitude-unresolved result, a raw effect narrowed by a stronger
control, a bounded aggregate that still preserves a mechanistic sub-finding, and a descriptive closeout that
does not rescue a failed confirmatory claim -- can these be represented consistently, without any transition
either (a) inventing evidence a defect actually destroyed, or (b) erasing evidence that legitimately survives
a narrower reinterpretation?

## Live check — do the ledger's own cited artifacts still exist?

`stage-01-artifacts.json` records, for every one of the 10 cases, which source files were found when the
ledger was originally built. Those paths were captured under an old repository root
(`C:\Projects\working-book\...`); this notebook rewrites that prefix to the current `REPO_ROOT` and checks
each path directly against this repository's filesystem -- a genuine executable check, not a re-citation.

```python
artifacts = json.loads((LEDGER_DIR / "stage-01-artifacts.json").read_text(encoding="utf-8"))
OLD_PREFIX = "C:\\Projects\\working-book\\"

resolved_rows = []
for case in artifacts["cases"]:
    n_found = 0
    n_total = len(case["paths"])
    for p in case["paths"]:
        rel = p.replace(OLD_PREFIX, "").replace("\\", "/")
        candidate = REPO_ROOT / rel
        if candidate.exists():
            n_found += 1
    resolved_rows.append({
        "case_id": case["case_id"],
        "declared_status": case["status"],
        "paths_cited": n_total,
        "paths_resolved_now": n_found,
        "all_resolve": n_found == n_total,
    })

resolved_df = pd.DataFrame(resolved_rows)
print(resolved_df.to_string(index=False))
print(f"\ncases with every cited artifact still resolving in this repo: "
      f"{int(resolved_df['all_resolve'].sum())} / {len(resolved_df)}")
```

## Re-deriving X003 live, from this notebook suite's own outputs

**X003 — "Stronger control narrows claim without erasing raw phenomenon"** links `CH28_V1_RAW`
(`validity=VALID`, `inferential_status=SUPPORTED`, mean 0.4402 against `SEI=0.15`) to `CH28_V2_EXCESS`
(`validity=VALID`, `inferential_status=BOUNDED_BELOW_SEI`, `evidence_role=NARROWS_CLAIM`, mean -0.0123
against `EXCESS_SEI=0.10`). The ledger's claim is specific and checkable: V2 does not invalidate or overturn
V1's raw measurement -- it narrows what that measurement is allowed to *mean*, while V1's `SUPPORTED` status
stands unchanged.

This reconstruction can check that relationship directly, because `16-causal-modularity-vs-matched-null.ipynb`
-- built earlier in this same notebook suite -- runs both stages of exactly this experiment live: a V1-style
raw `MODULE_SCORE` test at radius 4, then a V2-style excess-over-matched-null test. If that notebook exists
and has been executed, this cell parses its own stored outputs and checks whether *this reconstruction's own,
independently re-run* result reproduces the qualitative pattern X003 asserts -- not by re-citing the archived
`ch28` verdict a second time, but by reading what a different, already-built notebook in this same suite
actually computed. If Chapter 16's notebook is not available, this falls back to loading the raw
`ch28-causal-modularity-v1`/`v2` `stage-06-verdict.json` files directly, which is the un-upgraded version of
this same check.

```python
CH16_PATH = REPO_ROOT / "notebooks" / "16-causal-modularity-vs-matched-null.ipynb"

def extract_ch16_statuses(nb_path):
    '''Parse a notebook's *stored, already-executed* cell outputs and pull out the two
    status lines this notebook's own design prints: the V1 (H1) raw-modularity status and
    the V2 (H2) excess-over-matched-null status. Returns None if the file is missing or has
    no stored outputs (i.e. was never executed).'''
    if not nb_path.exists():
        return None
    nb_json = json.loads(nb_path.read_text(encoding="utf-8"))
    text_chunks = []
    for cell in nb_json.get("cells", []):
        if cell.get("cell_type") != "code":
            continue
        for out in cell.get("outputs", []):
            if out.get("output_type") == "stream":
                t = out.get("text", "")
                text_chunks.append(t if isinstance(t, str) else "".join(t))
    full_text = "\n".join(text_chunks)
    if not full_text.strip():
        return None

    h1 = re.search(r"H1 \(raw causal modularity exceeds SEI=[\d.]+\):\s*(\w+)", full_text)
    h2 = re.search(r"H2 \(excess causal modularity exceeds EXCESS_SEI=[\d.]+\):\s*(\w+)", full_text)
    h1_mean = re.search(r"V1 raw MODULE_SCORE at radius \d+:\s*mean=([\-\d.]+)", full_text)
    h2_mean = re.search(r"V2 excess MODULE_SCORE.*?mean=([\-\d.]+)", full_text)
    preserved = re.search(r"V1_status_preserved check: V1 was (\w+)", full_text)
    return {
        "source": "ch16_notebook_outputs",
        "h1_status": h1.group(1) if h1 else None,
        "h2_status": h2.group(1) if h2 else None,
        "h1_mean": float(h1_mean.group(1)) if h1_mean else None,
        "h2_mean": float(h2_mean.group(1)) if h2_mean else None,
        "v1_status_at_preservation_check": preserved.group(1) if preserved else None,
    }


live_result = extract_ch16_statuses(CH16_PATH)

if live_result is None:
    # Fallback: Chapter 16's notebook is not available or not executed -- re-derive directly
    # from the raw ch28 v1/v2 verdict JSON instead (the un-upgraded version of this check).
    print("Chapter 16 notebook not found or not executed -- falling back to raw ch28-* verdict JSON.")
    v1_verdict = json.loads((REPO_ROOT / "research" / "digital-life" / "ch28-causal-modularity-v1" / "stage-06-verdict.json").read_text(encoding="utf-8"))
    v2_verdict = json.loads((REPO_ROOT / "research" / "digital-life" / "ch28-causal-modularity-v2" / "stage-06-verdict.json").read_text(encoding="utf-8"))
    live_result = {
        "source": "ch28_raw_verdict_json_fallback",
        "h1_status": v1_verdict["primary_status"],
        "h2_status": v2_verdict["primary_status"],
        "h1_mean": None,
        "h2_mean": None,
        "v1_status_at_preservation_check": v1_verdict["primary_status"],
    }

print(json.dumps(live_result, indent=2))
```

```python
ledger_v1_row = ledger_df[ledger_df["case_id"] == "CH28_V1_RAW"].iloc[0]
ledger_v2_row = ledger_df[ledger_df["case_id"] == "CH28_V2_EXCESS"].iloc[0]

x003_check = next(c for c in consistency["cross_checks"] if c["check_id"] == "X003")

print(f"ledger CH28_V1_RAW.inferential_status:    {ledger_v1_row['inferential_status']}")
print(f"live re-derivation h1_status ({live_result['source']}):   {live_result['h1_status']}")
print(f"MATCH: {ledger_v1_row['inferential_status'] == live_result['h1_status']}")
print()
print(f"ledger CH28_V2_EXCESS.inferential_status: {ledger_v2_row['inferential_status']}")
print(f"live re-derivation h2_status ({live_result['source']}):   {live_result['h2_status']}")
print(f"MATCH: {ledger_v2_row['inferential_status'] == live_result['h2_status']}")
print()

x003_pattern_holds = (
    live_result["h1_status"] == "SUPPORTED"
    and live_result["h2_status"] == "BOUNDED_BELOW_SEI"
)
both_match_ledger = (
    ledger_v1_row["inferential_status"] == live_result["h1_status"]
    and ledger_v2_row["inferential_status"] == live_result["h2_status"]
)

print(f"X003 pattern ('stronger control narrows claim without erasing raw phenomenon') "
      f"holds under the live re-derivation: {x003_pattern_holds}")
print(f"Ledger's own claimed X003 pass flag: {x003_check['pass']}")
print(f"\nX003 RE-DERIVATION RESULT: "
      f"{'CONFIRMED -- this reconstruction independently reproduces the ledger cross-check' if (x003_pattern_holds and both_match_ledger) else 'DIVERGES -- see note below, this is reported honestly, not smoothed over'}")
```

## Controls and confounds

- **Frozen taxonomy, predeclared before classification** — the protocol's `forbidden_moves` list (printed
  above) was fixed before any of the 10 cases were classified: counting INVALID as evidence against a
  hypothesis, treating a CI crossing zero as FAILED, calling something bounded-negative without an explicit
  meaningful threshold, erasing valid sub-results when a larger inference is invalid, promoting a descriptive
  closeout into a confirmatory rescue, or silently changing the estimand after seeing the result.
- **Per-case rule gates, not a single aggregate score** — each of the 10 cases carries 7 independent boolean
  rule checks (`rule_invalid_not_negative`, `rule_bounded_requires_threshold`, etc.); "consistent" means every
  rule holds for every case, not that some average score clears a bar.
- **Artifact-path resolution checked live, not assumed** — the check above re-verifies, against this actual
  repository, that the ledger's cited source artifacts still exist rather than trusting the archived
  `stage-01-artifacts.json` claim at face value.
- **`manual_evidence_only_is_allowed: true`** — the protocol permits a case to rest on manually-recorded
  evidence if no machine artifact exists, but `stage-01-artifacts.json` reports `manual_evidence_only_count:
  0` for all 10 registered cases here — every case in this ledger rests on a discoverable artifact.
