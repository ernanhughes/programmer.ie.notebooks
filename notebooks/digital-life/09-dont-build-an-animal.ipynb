```python
EXPERIMENT = {"initial_memory": ["seed"], "growth_steps": 6}
EXPERIMENT
```

```python
state = {"history": ["seed"], "knowledge": set(), "storage": ["unit0"]}
trajectory=[]
for t in range(EXPERIMENT["growth_steps"]):
    state["history"].append(f"t{t}")
    state["storage"].append(f"unit{t+1}")
    trajectory.append(copy.deepcopy(state))
[(len(s["storage"]), list(s["history"])[-2:]) for s in trajectory]
```

## Checkpoint and restore

Process termination is not automatically informational death if enough state exists to continue.


```python
checkpoint = copy.deepcopy(trajectory[2])
continued = copy.deepcopy(trajectory[-1]); continued["knowledge"].add("late-discovery")
restored = copy.deepcopy(checkpoint); restored["knowledge"].add("alternate-discovery")
{"continued": continued, "restored_from_t2": restored}
```

## Forked continuation

Both branches inherit the same pre-fork history, then diverge.


```python
S = {"history": ["seed", "observe"], "knowledge": {"K"}}
S1 = copy.deepcopy(S); S2 = copy.deepcopy(S)
S1["history"].append("branch-A"); S1["knowledge"].add("A")
S2["history"].append("branch-B"); S2["knowledge"].add("B")
{"shared_prefix": S["history"], "branch_1": S1, "branch_2": S2}
```

## Bounded conclusion

The counterexamples show why biological mechanism does not equal universal computational requirement: persistence, forked continuation, and restored execution can be represented without reproduction, age, body, or death variables. They do not demonstrate digital life.
