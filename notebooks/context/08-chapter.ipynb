# Chapter 8 — Make the Window Bigger

## Question

**If context pressure is the problem, does a much larger window make context engineering unnecessary?**

Structural answer tested here: a larger window relieves capacity pressure while leaving every management question unchanged. Capacities below are fictional; computation curves are explicitly toy models. No model is benchmarked and no vendor is ranked.

## Setup — four concepts, two sides of one interface

Capacity, content, and representation are runtime-side context variables. Computation is model-side: what work the model performs over whatever arrives. The notebook manipulates the first three and labels the fourth without implementing it.

```python
from dataclasses import dataclass

SMALL_WINDOW = 8000    # fictional usable capacities, not vendor figures
LARGE_WINDOW = 20000

@dataclass(frozen=True)
class Bundle:
    tokens: int
    contradictions: int
    stale_markers: int
    scope_mismatches: int
    duplicates: int

bundle = Bundle(tokens=14000, contradictions=2, stale_markers=3,
                  scope_mismatches=1, duplicates=4)
print(f'bundle: {bundle.tokens} fixture tokens with deliberate defects.')
```

## Baseline — the bundle does not fit the small window

```python
print(f'small window {SMALL_WINDOW}: fits={bundle.tokens <= SMALL_WINDOW}')
assert bundle.tokens > SMALL_WINDOW, 'baseline must genuinely fail to fit'
```

## Intervention 1 — a larger window relieves capacity pressure

The obvious benefit, granted honestly: infeasible becomes feasible.

```python
print(f'large window {LARGE_WINDOW}: fits={bundle.tokens <= LARGE_WINDOW}')
assert bundle.tokens <= LARGE_WINDOW
```

## Intervention 2 — management problems survive the move

Capacity changed `fits?`. Nothing else moved. The defects below are counted, never resolved: resolution belongs to Chapters 19–20, 10, 21, and 5.

```python
before = bundle
after = Bundle(tokens=bundle.tokens, contradictions=bundle.contradictions,
                 stale_markers=bundle.stale_markers, scope_mismatches=bundle.scope_mismatches,
                 duplicates=bundle.duplicates)
for field in ('contradictions', 'stale_markers', 'scope_mismatches', 'duplicates'):
    print(f'{field:18s} before={getattr(before, field)} after={getattr(after, field)}')
    assert getattr(after, field) == getattr(before, field)
print('Capacity problem solved; context-management problems unchanged.')
```

## Positions are not capacity — four separate fields

Trained length, configured maximum, extrapolated reach, and effective usable context are different objects. The last is UNKNOWN here: it requires measurement of the Chapter 5 kind, and this notebook runs none.

```python
support = {'trained_length': 32768, 'configured_limit': 131072,
           'extended_limit': 1000000, 'effective_length': 'UNKNOWN'}
for k, v in support.items():
    print(f'{k:17s} {v}')
assert support['effective_length'] == 'UNKNOWN'
assert support['configured_limit'] != support['trained_length']
print('Configured maximum is not demonstrated usable context.')
```

## Toy computation curves — explicitly illustrative

TOY COMPLEXITY ILLUSTRATION — NOT PROVIDER PERFORMANCE. Dense interaction work grows disproportionately with length; sparser or bounded-state designs change the curve's shape. Values are toy units, not FLOPs, latency, memory, or cost, and no architecture is claimed to achieve any of them.

```python
lengths = [1000, 4000, 16000]
print(f"{'n':>6s} {'dense n^2':>10s} {'sparse n*k':>11s} {'bounded n':>9s} (toy units)")
for n in lengths:
    print(f'{n:6d} {n**2:10d} {n*64:11d} {n:9d}')
print('Different model-side curves leave runtime selection questions intact.')
```

## Boundary — two caches, not one

The prompt-prefix cache (serving layer, cross-request reuse) and the model-internal KV state (forward-pass internals, rebuilt per request) share vocabulary and differ in owner, granularity, and control. Chapter 9 owns the first; this chapter implements neither.

```python
prompt_prefix_cache = {'owner': 'serving layer', 'survives': 'across requests', 'protected_by': 'stable layout'}
kv_state = {'owner': 'model forward pass', 'survives': 'within one request', 'shrunk_by': 'model compression'}
assert prompt_prefix_cache['owner'] != kv_state['owner']
assert prompt_prefix_cache['survives'] != kv_state['survives']
print('Two ledgers. Pruning to save one routinely invalidates the other.')
```

## What this demonstrates

- Larger capacity can remove a hard-fit failure without resolving redundancy, scope, authority, staleness, representation, or usefulness.
- Model architecture and context management operate at different layers: one changes the cost curve, the other changes what is presented.

## What this does not demonstrate

- That any vendor or model is superior, or that any architecture achieves the toy curves.
- That configured length is effective usable length (marked UNKNOWN above).
- That sparse or recurrent designs preserve all information.
- That a bigger window harms behaviour, or that it removes the need for management.
