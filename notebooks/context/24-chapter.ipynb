## Intervention — help, hurt, unchanged, mechanically classified

````python
def classify(before, after):
    return 'HELPED' if after[1] > before[1] else ('HURT' if after[1] < before[1] else 'UNCHANGED')

classes = {t['id']: classify(t['before'], t['after']) for t in fb['tasks']}
for tid, cls in classes.items():
    print(f'{tid:20s} {cls}')
counts = {c: sum(1 for v in classes.values() if v == c) for c in ('HELPED', 'HURT', 'UNCHANGED')}
print(f"mechanical counts: {counts}; chapter aggregate: improved=2 degraded=1 unchanged=3")
assert counts == {'HELPED': 2, 'HURT': 1, 'UNCHANGED': 3}
print('No narrative smoothing: the failed negative control (calibration-echo) stays visible.')
````

## What this demonstrates

- Policy correctness, bundle correctness, and behavioural utility are three separable claims; the frozen rows pull each pair apart.
- Context helped, hurt, and did nothing on adjacent rows of one run: HELPED 2, HURT 1, UNCHANGED 3.
- The staged compiler earned construction machinery and demonstrated no additional behavioural utility over hard-gated greedy here.
- The earned architecture (small core, tripwired passes, outer evaluation loop) follows from the runs, not from preference.

## What this does not demonstrate

- That staged compilation never helps behaviourally: one matched set, one reader, six tasks.
- That conditional passes are unnecessary in general: each waits on its tripwire, not on dismissal.
- That the oracle failed, or that presence ever proved causality.
- Prevalence, external validity, or production benefit: zero genuine sessions.

## Connection — the series ends where it began

Chapter 1 asked how we would know whether changing the context actually mattered. The answer the discipline supports: fix everything but the bundle, vary the bundle, grade deterministically, refuse every unsupported inference.

> **Context engineering is not maximising how much an AI can see. It is deliberately constructing, tracing, and evaluating what this computation should see — while retaining the ability to discover that the decision was wrong.**
