## From channels to semantic channels

Iris follows conferences, labs, authors and feeds. A result appears outside her
channels and never uses the word "consensus." Her usual alerts do not find it.
She encounters it by accident three months later.

Subscribing to a relationship differs from subscribing to a source, creator, topic
or feed. A standing relational query names a kind of relation to a named anchor.


```python
subscriptions = ["a creator", "a publication", "a topic", "a feed", "a keyword"]
print("What people subscribe to now:")
for s in subscriptions:
    print(f"  - {s}")
print()
print("Finding material about a subject != finding evidence that changes a specified relationship.")
```

## Experiment: typed relation vs keyword match

A keyword alert for "consensus" misses the March paper that never uses the word.
A typed standing query for results bearing on a bounded-delay assumption could
surface it — but then a second job remains: determining whether the result
actually changes Iris's design.


```python
anchor = "bounded-delay assumption in Iris's protocol"
query_type = "typed relation: may weaken or qualify anchor"

print(f"Anchor: {anchor}")
print(f"Query: {query_type}")
print()
print("Stage 1: discover the March paper (not found by keyword alert).")
print("Stage 2: determine whether its result changes anything about Iris's design.")
print("Stage 2 cannot compensate for failure at Stage 1.")
```

## What this establishes

A semantic channel is a proposed architectural ambition, not an implemented
capability. Relation labels require scope and evidence; names alone do not make
them reliable.
