## Why this mechanism matters
Conceptual: ladder of representations from prose to formal specification.


```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## The language between languages: a ladder of representations

A product owner asks for a job queue: "A job must never be executed twice." A
worker times out after charging a card, and the caller retries. Does executed
mean the worker ran, the queue recorded completion, or money moved?

This notebook climbs the ladder from natural prose to formal specification,
showing what each form makes easier or harder.


```python
fixture = FIX["job_queue"]
print("Request:", fixture["request"])
print()
print("Readings of 'executed':")
for reading, meaning in fixture["readings"].items():
    print(f"  {reading}: {meaning}")
```

## Experiment: the ladder of representations

Each step on the ladder is mutually consistent, yet all can inherit the same initial
interpretive mistake. The missing step was asking the owner whether "never
executed twice" meant never charge twice.


```python
for i, step in enumerate(fixture["ladder"], 1):
    print(f"{i}. {step}")
```

## What this establishes

A specification can be internally consistent and still represent the wrong
obligation. The binding between option, definition and source must be declared
before checking. The ladder is a conceptual framework, not experimental results.
