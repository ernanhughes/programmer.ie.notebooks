## Compression has levels: progressive resolution

A candidate enters Iris's queue at three in the morning: a paper that may bear on
her protocol's timing assumption. The system must decide how much evidence the
candidate's initial representation carries.

Five proposed resolution levels: glance, brief, summary, detail, source. Each
defines what the reader can do, not a word count.


```python
levels = [
    ("GLANCE", "can I decide whether to inspect this candidate?"),
    ("BRIEF", "can I identify the reported change, its scope and standing?"),
    ("SUMMARY", "can I identify what to check before relying?"),
    ("DETAIL", "does it expose argument and evidence gaps?"),
    ("SOURCE", "the unchanged version"),
]

print("Resolution ladder (proposed interface states):")
for name, description in levels:
    print(f"  {name:8s}: {description}")
```

## Experiment: compression stop rule

Compression must stop when a required distinction cannot survive. A well-established
replication may need only identification and a route to the source. A possible
challenge to a design assumption needs its scope, evidence status and unanswered
questions visible even in a small preview.


```python
candidates = [
    {"id": "cp1", "status": "well-established replication", "resolution": "glance"},
    {"id": "cp2", "status": "possible challenge to timing assumption", "resolution": "summary"},
]

print("Resolution by candidate status (synthetic policy):")
for c in candidates:
    print(f"  {c['id']}: {c['status']:35s} -> {c['resolution']}")
```

## What this establishes

The resolution ladder is a proposed design, not a demonstrated psychological scale.
The reader may skip levels or open the source directly. Compression stops when a
required distinction cannot survive.
