## Semantic enhancement: reveal, re-express, add

A ten-year-old payments service has a forty-page architecture document. A retry
passage is dense prose. An enhancement appears beside the passage without
changing the canonical source.

Three acts differ in evidence requirements:
- **Reveal**: highlight structure already in the source.
- **Re-express**: render source information in another form (diagram, table).
- **Add**: bring in material not in the passage (external definition, link).


```python
passage = """Retry policy: three attempts total for non-idempotent requests. Idempotent requests follow the separate recovery path. Backoff doubles after each failure. Handling of third-party timeouts remains unresolved; consult design record D-17 before changing this policy."""

enhancements = {
    "reveal": "Highlight: attempts=3 (non-idempotent). Backoff doubles after each failure. Timeout handling = UNRESOLVED.",
    "re_express": "Table: attempt | status | backoff | 1 | running | 0s | 2 | failed→retry | 2s | 3 | failed→dead-letter | 4s",
    "add": "External: idempotency key definition from RFC 7231; see design record D-17 for timeout handling.",
}

print("Source passage:", passage)
print()
for kind, text in enhancements.items():
    print(f"--- {kind.upper()} ---")
    print(text)
    print()
```

## Experiment: source-supported vs externally added claims

The reveal and re-express enhancements are grounded in the passage. The add
enhancement brings in an external RFC definition that the passage never states.
An interface must label it as externally supplied, not as something the author
said.

A diagram arrow from retry 3 to a dead-letter queue would be an *addition* if the
source never specified that route. Re-expression must preserve uncertainty, not
fill blanks with plausible arrows.


```python
source_supported = {"reveal": True, "re_express": True, "add": False}
for kind, supported in source_supported.items():
    status = "supported by source" if supported else "external material — label separately"
    print(f"{kind:10s}: {status}")
```

## What this establishes

The boundary between source-derived and externally supplied material is the
chapter's central design constraint. Labels and provenance are the mechanism.
The boolean table is a synthetic illustration of the classification.
