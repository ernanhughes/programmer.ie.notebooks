## Protection that travels: portable communication boundary

Iris asks her reading tool to remind her before opening another paper. She
separately authorises a temporary change to its recommendation view. Those
permissions are local — they do not apply to her news or video services.

She then declares a portable rule: keep consensus-research recommendations out of
named feeds for this evening, with exceptions and an expiry.


```python
rule = {
    "principal": "Iris",
    "destinations": ["news service", "video service"],
    "feeds_affected": ["consensus research"],
    "exceptions": ["search", "direct opening"],
    "lifetime": "this evening, no later than 20:00",
    "carries_history": False,
}

print("Portable communication rule (constructed declaration):")
for key, value in rule.items():
    print(f"  {key}: {value}")
```

## Experiment: portability does not permit reinterpretation

Identifying the same user answers whose rule this is — not where it may operate.
A different service, broader topic filter or longer lifetime requires a different
decision. Portability does not permit reinterpretation.


```python
checks = {
    "rule exists": True,
    "user authorised use at specified destinations": True,
    "scope named (destinations, feeds, exceptions, lifetime)": True,
    "history not carried": True,
    "reinterpretation not permitted": True,
}

print("Portability conditions:")
for check, passed in checks.items():
    print(f"  {'PASS' if passed else 'FAIL'}: {check}")
```

## What this establishes

A communication policy can travel farther than the history that produced it, but
portability does not permit reinterpretation. This scene is constructed, not an
observed deployment.
