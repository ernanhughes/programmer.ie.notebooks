## How much of you can an AI reproduce? Four claims

The system forecasts Iris will choose the expanded calculation because she chose
it last time. That establishes something about the previous review, not about
this choice.

Four distinct claims: recall, estimation, prediction, reproduction. Prediction is
evidence about behaviour, not that the system became the person.


```python
claims = [
    ("RECALL", "what the person previously said or did"),
    ("ESTIMATION", "an inference about something presently unobserved"),
    ("PREDICTION", "a forecast of a specified future observable outcome"),
    ("REPRODUCTION", "a claim to reproduce behaviour across a stated range"),
]

print("Four distinct claims:")
for name, description in claims:
    print(f"  {name:12s}: {description}")
```

## Experiment: prediction does not become identity

A forecast must be registered before the outcome and scored on genuinely held-out
cases, compared with simple alternatives given the same context. Even an
accurate forecast supplies no permission to act.


```python
forecast = "Iris will choose the expanded calculation"
basis = "she chose the calculation in the previous review"

print(f"Forecast: {forecast}")
print(f"Basis: {basis}")
print()
print("Valid test requires:")
print("  - registered before the outcome")
print("  - scored on held-out cases")
print("  - compared with simple baselines")
print()
print("Even an accurate forecast: prediction != understanding != authority")
```
