## A model of me is not me

A colleague asks Iris to meet Tuesday. The system forecasts she will accept and
prepares a reply. Iris later says she would have accepted, and the draft matches
her wording — but the system was not authorised to send it.

Prediction concerning a bounded choice does not describe the whole person,
establish identity, or confer authority.


```python
chain = [
    "predicts a bounded choice",
    "!= describes the whole person",
    "!= establishes personal identity",
    "!= has authority to act for the person",
]

print("Four different claims, not levels on a scale:")
for link in chain:
    print(f"  {link}")
```

## Experiment: accurate prediction vs authorised action

The forecast and the wording can be correct while the send remains prohibited. A
system may have evidence about what Iris is likely to choose, but no evidence
from accuracy alone that she delegated the choice.


```python
prediction_correct = True
wording_matches = True
authorised_to_send = False

print(f"Prediction correct: {prediction_correct}")
print(f"Wording matches Iris's: {wording_matches}")
print(f"Authorised to send: {authorised_to_send}")
print()
print("Accuracy alone does not establish delegation.")
```

## What this establishes

A model may resemble Iris without being her, and may predict her without being
authorised to act for her. Acting for someone requires a grant for a particular
act.
