*Language: When Intelligence Enters the Channel* — Part II — Too Much Information

This notebook is a practical companion to Chapter 16. It is not an abbreviated version of the chapter; it demonstrates the chapter's mechanism with executable, offline code.

## What you will learn
- That expertise in a subject does not establish familiarity with a source's assumptions.
- That personal novelty is about specific capability, not a stable profile.
- That removing an explanation can remove the qualification needed to interpret.

## Why this mechanism matters
Context: personal novelty vs salience vs source necessity vs task relevance.


```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## Don't tell me what I already know: personal novelty

Iris knows consensus systems but not this speaker's particular timing assumption.
The system removes the definition of partial synchrony, assuming she knows it.
She sees the performance claim without its local definition and overstates the
guarantee.

Expertise in a subject does not establish familiarity with a particular source's
assumptions.


```python
fixture = FIX["personal_state"]
print("Declared:")
for d in fixture["declared"]:
    print(f"  - {d}")
print("Observed:")
for o in fixture["observed"]:
    print(f"  - {o}")
print("Inferred:")
for i in fixture["inferred"]:
    print(f"  - {i}")
```

## Experiment: three models are not interchangeable

A model of what you know is not a model of what you like or who you are. Iris's
clicks on tables do not establish that she prefers tabular views across all
tasks. A click is not a stable preference.


```python
models = ["model of what you know", "model of what you like", "model of who you are"]
print("Three easily confused models (distinct):")
for m in models:
    print(f"  - {m}")

print()
print("Iris's expertise in consensus research does not answer:")
print("  - Has she seen this speaker's definition?")
print("  - Does she know which timing assumption this speaker chose?")
```

## What this establishes

The model should say what evidence exists for a specific capability, not generalise
from expertise to a kind of person. The three model types are a conceptual
distinction, not experimental classifications of Iris.
