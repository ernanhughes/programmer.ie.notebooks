## Capstone III — The Personal AI

Iris begins an architecture review with ten minutes. Six mechanisms or records are
available: standing policy, observations, provisional inference, temporary view,
two authorised boundaries. Coexistence is not a reason to consult them all.

The question: how much personal state does this communication actually require?


```python
inputs = [
    "standing policy (declared)",
    "observations from earlier reviews",
    "provisional inference about derivations",
    "temporary view (session-scoped)",
    "authorised boundary 1",
    "authorised boundary 2",
]

print("Six available inputs, not six necessities:")
for i, inp in enumerate(inputs, 1):
    print(f"  {i}. {inp}")
```

## Experiment: more personal state does not mean better communication

Before retrieving accumulated personal state, the system asks which part is needed
for Iris's current request. Retrieving everything first exposes more personal
information than the task may need.

Each additional claim must justify its collection, retention and use.


```python
task = "show material bearing on the decision; keep equations visible; fold supporting steps"
needed_inputs = ["standing policy", "temporary view", "authorised boundary 1"]

print(f"Task: {task}")
print(f"Inputs actually needed: {needed_inputs}")
print(f"Inputs not consulted: {len(inputs) - len(needed_inputs)} of {len(inputs)}")

print()
print("more personal state != better communication")
```

## What this establishes

The central design question is not whether the system can accumulate a personal
model, but whether each additional claim improves an authorised communication
decision enough to justify collecting, retaining and using it.
