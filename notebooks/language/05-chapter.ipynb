```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## Visual compression: previews for selection decisions

A security engineer faces forty browser tabs with an hour before stand-up. She
will open perhaps six. Each tab's title, favicon and first lines are a bet:
open now, defer, or skip. A preview is a smaller representation derived from a
source for a declared task.

The two preview failures are false open (looks relevant, wastes time) and false
skip (looks irrelevant, hides critical evidence).


```python
candidates = FIX["subscription_candidates"]
print("Candidate previews (synthetic, constructed):")
for c in candidates:
    print(f"  {c['candidate_id']}: {c['title']}")
    print(f"    relation={c['relation']}  relevance={c['task_relevance']}  novelty={c['novelty']}")
```

## Experiment: false open vs false skip under a budget

Suppose Iris's attention budget is two items. A preview policy ranks by
task_relevance and surfaces c1 and c4. c2 (a major library update) is skipped —
potentially a false skip if it contains a security patch. c3 (familiar survey)
is correctly deferred.

The preview improves selection speed but cannot know the value of unopened
sources. Selection quality needs a task-specific reference judgement.


```python
budget = 2
ranked = sorted(candidates, key=lambda c: {"critical": 3, "high": 2, "medium": 1, "low": 0}[c["task_relevance"]], reverse=True)
opened = ranked[:budget]
skipped = ranked[budget:]

print(f"Attention budget: {budget} items")
print("Opened:", [c["candidate_id"] for c in opened])
print("Skipped/deferred:", [c["candidate_id"] for c in skipped])

false_skip_risk = [c for c in skipped if c["task_relevance"] == "critical"]
print(f"Potential false skips: {[c["candidate_id"] for c in false_skip_risk]}")
```

## What this establishes

The preview policy opens the two highest-relevance items and defers the rest. A
critical item can still be skipped if relevance scoring is wrong — that is the
false-skip risk. The counts and rankings are synthetic demonstrations, not
measured selection outcomes.
