## Important to whom? Task relevance as a relation

Three colleagues ask about the same consensus-protocol talk for different reasons:
Iris needs safety under unbounded delays, Tom compares latency claims, Mina wants
to reproduce the demonstration. The source and units are the same; the selected
bundles need not be.

Salience, source necessity, task relevance and personal novelty are separate
reasons something might look important.


```python
reasons = ["salience", "source necessity", "task relevance", "personal novelty"]
print("Four reasons a unit might look important (distinct):")
for r in reasons:
    print(f"  - {r}")
print()
print("salience != source necessity != task relevance != personal novelty")
```

## Experiment: relevance depends on the declared purpose

With the same source and attention budget, Iris's bundle includes the minute-53
qualification (safety under unbounded delay). Tom's bundle includes the
minute-42 demonstration (latency claims). Mina's bundle includes both plus the
fault definition. Relevance is a relation among unit, purpose and context.


```python
bundles = {
    "Iris (safety under unbounded delay)": ["u3", "u1"],
    "Tom (latency comparison)": ["u2"],
    "Mina (reproduce demonstration)": ["u1", "u2", "u3"],
}

for person, unit_ids in bundles.items():
    print(f"  {person:40s} -> units {unit_ids}")
```

## What this establishes

The same source yields different bundles for different tasks. Source necessity
constrains what a selected claim may leave behind. These bundles are synthetic
illustrations of the mechanism, not measured reader outcomes.
