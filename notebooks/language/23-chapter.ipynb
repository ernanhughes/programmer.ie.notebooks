## From observation to policy: memory can be right and still make the present wrong

In March Iris says the old replication path is gone and writes go through quorum
path B. In June the service migrates to path C. The March record remains
accurate evidence of what she said, but the answer it produces in October is
wrong.

Memory needs scope, time and applicability — not just accuracy.


```python
timeline = [
    {"when": "March", "event": "Iris says writes go through quorum path B", "accurate_at": True},
    {"when": "June", "event": "Migration replaces path B with path C", "accurate_at": True},
    {"when": "October", "event": "System answers from March memory", "answer": "wrong — architecture changed"},
]

print("Timeline (constructed case):")
for entry in timeline:
    print(f"  {entry['when']}: {entry['event']}")
    if "answer" in entry:
        print(f"        Answer: {entry['answer']}")
```

## Experiment: scope and time govern memory use

The March statement had a speaker, subject, scope and time. Recording it is not
independently verifying the architecture. Policy promotion from observation
should always be a question, not an automatic consequence.


```python
def memory_applicable(record, current_context):
    """Toy check: is this memory still applicable to the current context?"""
    return record["scope"] == current_context["service"] and record["time"] <= current_context["now"]

record = {"scope": "quorum path B", "time": "March"}
context_now = {"service": "quorum path C", "now": "October"}

print(f"Memory applicable to current context: {memory_applicable(record, context_now)}")
print("Reason: the service changed from path B to path C.")
```

## What this establishes

Accurate recall can lead the system astray when the world changes. Memory must carry
its scope and governing conditions. This is a constructed case, not a measured
failure.
