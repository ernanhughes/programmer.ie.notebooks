## The Taint we leave: a co-produced hypothesis

Two people independently authorise a private search. Both repeatedly ask for the
qualification behind a headline. Both reject polished answers without traceable
evidence. Both turn to counterexamples. The system proposes a connection for a
possible collaborator without naming anyone.

Taint is a proposed machine-detectable pattern co-produced by a person and the
models, prompts, tools, tasks and corrections across repeated interactions.


```python
fixture = FIX["language_runtime_trace"]
print("Taint (proposed definition):")
print("  a potentially machine-detectable, context-dependent pattern co-produced")
print("  by a person and the models, prompts, tools, tasks, corrections and")
print("  recording practices present across repeated interactions.")
print()
print("Status: hypothesis. No detector built or evaluated for this book.")
```

## Experiment: matching ≠ reveal ≠ introduction

The proxy offers a reason to consider a possible collaborator without exposing the
other person's identity or history. Match existence can itself leak information.
No stage authorises the next.


```python
boundary = [
    "matching ≠ reveal",
    "reveal ≠ introduction",
    "similarity score does not establish a relationship",
]

print("Boundaries that must remain visible:")
for b in boundary:
    print(f"  - {b}")
```

## What this establishes

Taint is neither a necessary component of the language thesis nor proof that the
channel knows the person. If a detectable pattern cannot outperform simpler
explanations (topic, task, writing style, shared provider), the term need not
become a software object.
