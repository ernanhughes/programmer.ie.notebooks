## Sometimes the best interface adds friction

Iris says: for this session, finish this paper first, remind me before opening
another. A new result appears through her subscription. The interface puts a
reminder in her path: *You asked for a reminder before opening another paper.*
She can either continue or open the new paper.

The system may add friction only in service of an objective the user actually
supplied. Support is not control.


```python
intervention = {
    "user_goal": "finish this paper first",
    "trigger": "opening another paper during this session",
    "intervention": "reminder line with two equal choices",
    "override": "Iris can open the new paper without a second confirmation",
    "scope": "this session only",
}

print("Friction intervention (constructed scene):")
for key, value in intervention.items():
    print(f"  {key}: {value}")
```

## Experiment: user-authored goal vs system-inferred benefit

The system's authority comes from the intervention Iris requested, not from a
judgement that finishing the first paper would be better for her. A standing
subscription authorising discovery is not permission to obstruct navigation.


```python
checks = {
    "objective user-supplied": True,
    "user authorised type/scope/duration": True,
    "override available": True,
    "applicability re-established": True,
}

print("Necessary conditions for adding friction:")
for check, passed in checks.items():
    print(f"  {'PASS' if passed else 'FAIL'}: {check}")

print()
print("All conditions met → reminder may be shown. Iris retains override.")
```

## What this establishes

Friction is a contract the user wrote, with a visible source, easy override and clear
expiry. It is never a judgement the system makes for the user. This is a
constructed scene, not evidence that reminders improve concentration.
