## There is no universal best interface

Iris opens tables more often than prose. A system could turn these traces into a
stable label: "this person is a diagram person." But the trace has no record of
what alternatives were available, why she opened each view, or whether she
performed better.

The claim under examination: a stable representational type reliably matches
format to performance. The evidence standard is a crossover, not a main effect.


```python
print("Learning-style matching requires a crossover:")
print("  Type A: representation A > representation B")
print("  Type B: representation B > representation A")
print()
print("A main effect (everyone does better with diagrams) cannot justify matching by style.")
```

## Experiment: stated preferences vs task demands vs accessibility

Stated preferences, task strategies, prior knowledge and accessibility requirements
are different inputs with different evidential requirements. None should be quietly
converted into a general-purpose category.

Pashler et al. (2008) found no adequate evidence base for learning-style matching.
Clinton-Lisell and Litzinger (2024) found only about 26% of outcomes showed the
crossover pattern.


```python
inputs = ["stated preferences", "task demands", "prior knowledge evidence", "accessibility constraints"]
print("Different inputs, not one category:")
for inp in inputs:
    print(f"  - {inp}")

print()
print("Crossover outcomes supporting matching: ~26% (Clinton-Lisell & Litzinger 2024, literature)")
print("These are educational results, not AI-interface measurements.")
```

## What this establishes

Rejecting the learning-style shortcut does not reject individual differences. It
means each input must be evaluated separately. The chapter examines the shortcut,
not the existence of individual differences.
