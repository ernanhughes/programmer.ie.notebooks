## Why this mechanism matters
Memory: declared / observed / inferred persistence origins.


```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## The personal experience: declared / observed / inferred

Iris notices the Radar behaves differently over time. One rule came from something
she explicitly said. Another may have come from a repeated action the system
observed. A third may be a conclusion the system drew. All three change a view;
they do not have the same authority.

Three persistence origins: declared (explicitly supplied), observed (authorised to
record), inferred (proposed from evidence).


```python
fixture = FIX["personal_state"]
origins = ["declared", "observed", "inferred"]
print("Three persistence origins with different authority:")
for origin in origins:
    items = fixture[origin]
    print(f"  {origin.upper()}: {len(items)} items")
    for item in items:
        print(f"    - {item}")
```

## Experiment: a click is not a stable preference

Iris opens tables more often than prose. That observed behaviour could be
classified as "prefers tables" — but the trace does not record what alternatives
were available, why she opened each view, or whether she performed better.

Declared instructions carry more authority than inferred interpretations.


```python
authority = {
    "declared": "high — Iris explicitly supplied",
    "observed": "medium — authorised to record, needs context",
    "inferred": "low — system's interpretation, may be wrong",
}

print("Authority by origin (conceptual):")
for origin, level in authority.items():
    print(f"  {origin:10s}: {level}")
```

## What this establishes

A remembered preference can be useful and still be wrong, stale or unauthorised.
Memory must show its origin, scope and governing permission. These are Part III
design principles, not measurements of Iris.
