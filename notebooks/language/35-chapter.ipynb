## Private discovery between people: five stages, five permissions

Iris wants to find a qualified collaborator for an unpublished protocol failure
without making it public. The proposed protocol separates matching, consent for
a particular candidate, identity reveal, introduction and disclosure.

Match existence can itself leak information. No stage authorises the next.


```python
stages = [
    (1, "purpose-authorised matching", "consent to be eligible"),
    (2, "independent scoped consent for the candidate", "consent to proceed to identification"),
    (3, "approved identity reveal", "separate permission"),
    (4, "separately allowed introduction", "separate permission"),
    (5, "separately approved evidence disclosure", "separate permission"),
]

print("Five discovery stages, each with distinct permission:")
for num, stage, permission in stages:
    print(f"  Stage {num}: {stage}")
    print(f"        Permission: {permission}")
```

## Experiment: no stage authorises the next

Discovery should precede discretionary disclosure. Permission to match must precede
discovery itself. Telling Iris "someone matched" can leak another person's
participation.


```python
chain = "consent to match → purpose-limited match → consent per candidate → identity reveal → introduction → disclosure"
print("Proposed chain (not a demonstrated privacy guarantee):")
print("  ", chain)

print()
print("Each transition has a distinct audience, information release, authority check")
print("and failure condition. No stage authorises the next.")
```

## What this establishes

The proposal is a sequence of separately checked operations, not a guarantee that all
useful matching can be done privately. This chapter is about boundaries, not a
built system.
