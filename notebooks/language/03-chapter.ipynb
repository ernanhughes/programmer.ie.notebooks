## Prose vs diagram vs demonstration

A senior colleague explains a distributed-systems deadlock in three forms:
prose, a timeline diagram on a whiteboard, and an acted-out demonstration with
coffee cups as locks. Each form makes different relationships available.

This notebook simulates the three representations and shows what a receiver can
reconstruct from each.


```python
fixture = FIX["deadlock"]
print("Source:", fixture["source"])
print()
for form, text in fixture["representations"].items():
    print(f"--- {form.upper()} ---")
    print(text)
    print()
```

## Experiment: what each representation makes visible

Prose states the circular wait precisely but requires the reader to hold both
services and both locks in working memory. The diagram makes the circular
relationship visible at a glance. The demonstration adds temporal and physical
grounding but is transient.

None contains different facts; each imposes different reconstruction work.
The chapter's point is that form changes what can be recovered, not just how
it looks.


```python
relations = {
    "prose": ["A holds lock 1", "A waits for lock 2", "B holds lock 2", "B waits for lock 1"],
    "diagram": ["A holds lock 1", "A waits for lock 2", "B holds lock 2", "B waits for lock 1", "circular wait visible"],
    "demonstration": ["A holds lock 1", "A waits for lock 2", "B holds lock 2", "B waits for lock 1", "circular wait visible", "temporal sequencing"]
}
for form, items in relations.items():
    print(f"{form:14s} -> {len(items)} recoverable elements")
```

## What this establishes

The count of recoverable elements grows from prose to demonstration. These are
illustrative counts, not experimental measurements. The chapter's caution is
that the demonstration came after two earlier explanations — format alone did
not cause the improvement.
