## What you will learn
- That speech is typed by the act: informing, requesting, offering, accepting.
- That preparation (draft) is not transmission, and transmission is not the speech act.
- That the same words may perform different acts in different settings.

## Why this mechanism matters
Authority: speech acts typed by the act; preparation vs transmission.


```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## What may my proxy say for me? Speech acts typed by the act

A hotel counteroffers €400/night with a free upgrade. Iris's grant permits one
refundable room up to €350/night — it does not authorise accepting €400. The
system prepares "€400 works for me." Preparing words is not transmitting them.

Speech is typed by the act: informing, requesting, offering, accepting, promising.


```python
scenario = FIX["proxy_scenario"]
print("Principal:", scenario["principal"])
print("Grant:", scenario["grant"])
print("Counteroffer:", scenario["counteroffer"])
print("Prepared message:", scenario["prepared_message"])
print()
print("Speech acts:")
for act in scenario["speech_acts"]:
    print(f"  - {act}")
```

## Experiment: preparation vs transmission vs speech act

The same four words may be a private draft, a personally approved sending, or an
apparent acceptance transmitted without permission. The words alone do not settle
what happened — sender, recipient, exchange and terms contribute to the act's
force.


```python
settings = [
    "private draft — Iris reads and rejects",
    "personally approved and sent by Iris",
    "authorised negotiator sends as part of agreed exchange",
    "system transmits apparent acceptance without permission",
]

print("Possible settings for the same words:")
for s in settings:
    print(f"  - {s}")

print()
print("Gate requirement: proposed speech act, audience, authority, and actual text.")
```

## What this establishes

Calling a transmission a negotiating probe does not stop the hotel from treating it
as assent. The channel must account for the act and its plausible consequences
before sending. This chapter is about authority typed by the act.
