## The fixed page: one representation for many readers

Three people consult the same twenty-page clinical guideline: a paramedic, a
junior doctor and a medical student. They need different routes through the
same document. The fixed page was chosen before any of them arrived.

This notebook shows one source supporting multiple task-specific views without
losing obligations, qualifications or authority.


```python
source = """CLINICAL GUIDELINE v3.2 — Suspected Stroke

1. Triage: assess FAST signs (Face, Arms, Speech, Time).
2. Escalation pathway: stroke team within 10 minutes.
   Exception: if glucose < 3.0, treat hypoglycaemia first.
3. Treatment: thrombolysis within 4.5 hours of onset.
4. Contraindications: recent surgery, active bleeding, anticoagulation.
5. Version: 3.2, effective 2026-01-15, locally authorised."""

views = {
    "paramedic": "Triage → escalate to stroke team in 10 min. If glucose < 3.0, treat hypoglycaemia first.",
    "junior_doctor": "Thrombolysis within 4.5 h of onset. Contraindications: recent surgery, active bleeding, anticoagulation. v3.2.",
    "medical_student": "Why the thresholds exist: FAST signs identify large-vessel occlusion; 4.5 h is the thrombolysis window.",
}

print("Source version: guideline v3.2")
print()
for role, view in views.items():
    print(f"--- {role} ---")
    print(view)
    print()
```

## Experiment: what each view preserves and what it omits

The paramedic view preserves triage and escalation but omits the 4.5-hour window and
contraindications — appropriate for ambulance use, dangerous if used for treatment
decisions. The junior doctor view preserves treatment and contraindications but not
the pedagogical rationale. The student view preserves rationale but not actionable
steps.

One source does not have one universally best rendering. Each view must preserve
the obligations relevant to its task.


```python
obligations = {
    "paramedic": ["triage", "escalation pathway"],
    "junior_doctor": ["treatment instructions", "contraindications", "version-specific cautions"],
    "medical_student": ["why thresholds exist", "pathway rationale"],
}
source_topics = ["triage", "escalation", "glucose exception", "thrombolysis", "4.5 hours",
                   "contraindications", "version 3.2"]

for role, needed in obligations.items():
    covered = [t for t in needed if any(t.lower() in views[role].lower() for _ in [0])]
    missing = [t for t in needed if t not in covered]
    print(f"{role:14s} covered={covered} missing={missing}")
```

## What this establishes

Each task-specific view covers its required topics and omits others. The omissions
are appropriate for the task but would be failures if the view were used for a
different task. Reader-time rendering must declare which obligations it preserves.
