## What you will learn
- That identical words can be reconstructed differently by different receivers.
- That prior knowledge changes what is recovered, not just what is read.
- That reconstruction is work the receiver performs, not meaning delivered.

## Why this mechanism matters
Conceptual: identical representation, different reconstructions based on prior knowledge.


```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## Identical representation, different reconstructions

Two registrars receive the same note about bed seven: *keep an eye on him*.
One covered the ward last week and infers a specific action (check within the
hour). The other is new and infers only general vigilance. Both read the words
accurately. The note alone cannot establish which reconstruction matches the
sender's intent.


```python
fixture = FIX["ward_handover"]
print("Representation:", fixture["representation"])
print()
for rec in fixture["reconstructions"]:
    print(f"--- {rec['receiver']} ---")
    print(f"Prior knowledge: {rec['prior_knowledge']}")
    print(f"Reconstruction: {rec['recovered']}")
    print()
```

## Experiment: what came from the message vs what came from prior knowledge

The experienced registrar's reconstruction includes a one-hour deadline that does
not appear in the representation. That deadline came from prior knowledge, not
from the words. The new registrar's reconstruction lacks the urgency entirely.

Neither reconstruction can be validated from the representation alone. This is
why every mediated representation must be judged against what a particular
receiver can recover, not against the sender's fluency.


```python
rep = fixture["representation"]
for rec in fixture["reconstructions"]:
    recovered_words = set(rec["recovered"].lower().split())
    rep_words = set(rep.lower().split())
    from_rep = recovered_words & rep_words
    not_from_rep = recovered_words - rep_words
    print(f"{rec['receiver']}")
    print(f"  words also in representation: {sorted(from_rep)}")
    print(f"  words NOT in representation: {sorted(not_from_rep)}")
    print()
```

## What this establishes

The word-overlap analysis shows which recovered words are grounded in the
representation and which came from the receiver's prior knowledge. These are
synthetic demonstrations of the mechanism; real comprehension requires reader
studies that this book does not report.
