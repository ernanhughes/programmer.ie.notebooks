## From conversation to delegation: explicit grants

Iris delegates one bounded operation: book one refundable hotel within limits.
The grant must specify price, distance, dates and the meaning of refundable.
Any missing consequential term is a reason to clarify, not a blank for the model
to fill.

Delegation is an explicit, inspectable grant — not the system's interpretation of
usefulness.


```python
grant = {
    "operation": "book one refundable hotel",
    "limits": {"price_per_night_max": 350, "currency": "EUR"},
    "dates": "specified",
    "refundable": "required",
    "return_purchase": False,  # not unless outside terms
    "data_use": "permitted during planning and execution",
    "revocable": True,
}

print("Explicit grant (constructed):")
for key, value in grant.items():
    print(f"  {key}: {value}")
```

## Experiment: grant narrows authority and supports revocation

A prepared booking is not a completed booking. Before any submission, the channel
must identify both the operation Iris authorised and the exact operation it is
about to perform. Receipts and outcome reconciliation make delegation
inspectable.


```python
pipeline = [
    "desired participation (which steps Iris wants to retain)",
    "explicit, inspectable grant (for this actor and operation)",
    "permitted data use during planning and execution",
    "operation-bound authorisation and shared-limit check",
    "external attempt and outcome reconciliation",
]

print("Delegation pipeline:")
for i, step in enumerate(pipeline, 1):
    print(f"  {i}. {step}")
```

## What this establishes

Desired participation guides a grant but is not itself permission for every account
access, disclosure or payment. A standing grant is not a reason to disregard a
more restrictive current instruction.
