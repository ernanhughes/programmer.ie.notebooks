## Enhanced interaction: the sidecar

A doctoral student reads a paper citing "the well-known impossibility result"
without proof or reference. A sidecar beside the paper can surface the original
theorem, check its assumptions, and preserve her reading place.

Three labels recur: **derived** (re-expressed from the source), **retrieved**
(found outside the source), and **interactive** (reader-requested). Origin and
interaction are different questions.


```python
items = [
    {"id": "s1", "origin": "derived", "interactive": False, "label": "Claim map re-expressed from this paper"},
    {"id": "s2", "origin": "retrieved", "interactive": False, "label": "Original theorem from cited paper"},
    {"id": "s3", "origin": "retrieved", "interactive": True, "label": "Reader asked: which assumptions must hold?"},
]

print("Sidecar items (constructed illustration):")
for item in items:
    print(f"  {item['id']}: origin={item['origin']:10s} interactive={item['interactive']}")
    print(f"         {item['label']}")
```

## Experiment: provenance labels for sidecar items

A derived claim map needs source spans and an interpretation check — a label saying
"from this document" is not enough. A retrieved theorem creates a candidate and a
source trail, not a verified fact. An interactive response may combine all three.

The reader's request authorises an attempt to answer; it does not grant the answer
authority.


```python
for item in items:
    if item["origin"] == "derived":
        req = "source spans + interpretation check"
    elif item["origin"] == "retrieved":
        req = "external source trail + version + passage"
    else:
        req = "combination; each component independently accountable"
    print(f"{item['id']}: requires {req}")
```

## What this establishes

The sidecar is an optional presentation surface, not a second author. Each item must
remain independently accountable to its origin. The three-way classification is
the chapter's design vocabulary, not experimental results.
