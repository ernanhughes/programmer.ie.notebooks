## What do I want to keep? Capability vs desire vs permission

A meeting request arrives. The system can notice it, interpret it, check a
calendar, suggest a reply and send one. Iris says: check when she's free and
prepare a reply; she will decide whether the meeting belongs in the week.

Capability, desire and permission are three different questions.


```python
division = [
    ("capability", "can the system perform the step?"),
    ("desire", "does Iris want the system to take this role?"),
    ("permission", "is the operation authorised for this actor, here and now?"),
]

print("Three separate questions:")
for name, description in division:
    print(f"  {name:10s}: {description}")
print()
print("capability != desire != permission")
```

## Experiment: Iris keeps judgement and sending

Iris wants help gathering information and preparing a response. She keeps the
judgement about whether to meet, the commitment about time, and the decision to
send. A draft saying "Tuesday works" could presume the answer she reserved.


```python
iris_keeps = ["judgement whether to meet", "commitment about time", "decision to send"]
system_helps = ["check availability", "prepare a reply draft"]

print("Iris keeps:")
for item in iris_keeps:
    print(f"  - {item}")
print("System helps:")
for item in system_helps:
    print(f"  - {item}")
```

## What this establishes

The answers to capability, desire and permission can differ. A previously granted
permission is not a reason to disregard a more restrictive current instruction.
This chapter concentrates on the desired division of work.
