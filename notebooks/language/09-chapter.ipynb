## Similarity is not discovery

A retrieval system ranks a paper first because it shares vocabulary with the reader's
paper. Same benchmarks, same metrics, same dataset family — but it contributes
nothing to the reader's question. Similarity is only one relation, and it is not
the relation that usually matters.

This notebook computes a synthetic similarity matrix showing that contradiction and
paraphrase can be close under vector distance.


```python
import numpy as np

# Synthetic embeddings for demonstration only — not real model embeddings.
np.random.seed(FIX["seed"])

texts = [
    "The treatment improved outcomes",
    "The treatment did not improve outcomes",  # contradiction, shares words
    "Patient outcomes improved with the treatment",  # paraphrase
    "Unrelated topic: quantum computing algorithms",
]

# Deterministic synthetic vectors (not from any model)
vectors = np.random.randn(len(texts), 8)
vectors = vectors / np.linalg.norm(vectors, axis=1, keepdims=True)

similarity = vectors @ vectors.T

print("Cosine similarity matrix (synthetic, not model embeddings):")
print(np.round(similarity, 2))
print()
for i in range(len(texts)):
    for j in range(i+1, len(texts)):
        print(f"  {i} ↔ {j}: similarity={similarity[i,j]:.2f}")
```

## Experiment: contradiction can be similar to affirmation

Text 0 and text 1 share almost every word but disagree about the outcome. Depending
on the embedding model and corpus, their vectors may be close. Text 0 and text 2
are a paraphrase. A high similarity score does not say whether one text supports,
refutes or explains the other.

The interface must name the relation it exposes — similarity is not discovery.


```python
# Find the highest off-diagonal similarity
mask = ~np.eye(len(texts), dtype=bool)
i, j = np.unravel_index(np.argmax(similarity * mask), similarity.shape)
print(f"\nHighest similarity: texts {i} and {j} ({similarity[i,j]:.2f})")
print(f"  Text {i}: {texts[i]}")
print(f"  Text {j}: {texts[j]}")
print("  Relation: contradiction — similarity alone cannot establish this.")
```

## What this establishes

The similarity matrix is a synthetic demonstration with randomly generated vectors.
It shows that proximity under one measurement is not verification of a semantic
relation. No claim about any real embedding model is made.
