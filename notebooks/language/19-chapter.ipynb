## When should the AI speak? Delivery timing

A candidate arrives at 14:40: a paper that may bear on an assumption behind code
being deployed at 18:00. Iris has a design review until 15:30. The default is to
queue until the first break. Interruption is a judgement, not a fact.

Delivery is a separately auditable, permission-bound decision under uncertainty,
freshness and deadline constraints.

## Experiment: delivery under constraints

Before delivery the system checks currency, authorisation and interval remaining. If
any check fails, the policy must choose again or say it cannot establish a safe
delivery time. The reader may correct or reopen decisions when evidence changes.


```python
now = "14:40"
review_until = "15:30"
deployment = "18:00"
candidate_current = True
iris_authorised = True
interval_sufficient = True

checks = {
    "candidate is current": candidate_current,
    "Iris is authorised to receive": iris_authorised,
    "interval remaining is sufficient": interval_sufficient,
}

print(f"Current time: {now}")
print(f"Design review until: {review_until}")
print(f"Deployment: {deployment}")
print()
for check, passed in checks.items():
    print(f"  {'PASS' if passed else 'FAIL'}: {check}")

all_pass = all(checks.values())
print()
if all_pass:
    print("Decision: queue until first suitable break (~15:35).")
else:
    print("Decision: cannot establish a safe delivery time.")
```

## What this establishes

The timing judgement is constructed for illustration. Relevance, qualification and
resolution are inputs, not settled truths. The system must be able to say when it
cannot establish a safe delivery time.
