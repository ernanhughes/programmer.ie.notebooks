```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## The three joints of communication

A registrar hands over a ward of twenty patients in ten minutes. What she holds
in mind is richer than her message. This notebook separates the three joints:
**intent** (what she wants the day team to recover), **representation** (the
words, numbers and gestures she actually transfers), and **reconstruction** (what
the day team rebuilds from those materials plus what they already know).


```python
fixture = FIX["ward_handover"]
print("Intent:", fixture["intent"])
print()
print("Representation:", fixture["representation"])
print()
for rec in fixture["reconstructions"]:
    print(f"Receiver: {rec['receiver']}")
    print(f"  Prior knowledge: {', '.join(rec['prior_knowledge'])}")
    print(f"  Recovered: {rec['recovered']}")
    print(f"  Added (not in representation): {rec['added']}")
    print(f"  Missed: {rec['missed']}")
    print()
```

## Experiment: the same words, two reconstructions

The representation is identical for both receivers. What differs is what each
receiver brings. The experienced registrar infers a one-hour deadline that was
never in the words; the new registrar does not. The first adds something (the
deadline) that was not represented; the second misses the urgency.

This demonstrates the chapter's first principle: **the representation is not the**
**meaning**. The receiver rebuilds meaning from the evidence the representation
supplies, using whatever prior knowledge they have.


```python
import matplotlib.pyplot as plt

receivers = [r["receiver"] for r in fixture["reconstructions"]]
recovered_counts = [len(r["recovered"].split()) for r in fixture["reconstructions"]]
added_counts = [len(r["added"].split()) for r in fixture["reconstructions"]]

fig, ax = plt.subplots(figsize=(8, 4))
x = range(len(receivers))
ax.bar([i - 0.2 for i in x], recovered_counts, width=0.4, label="words recovered")
ax.bar([i + 0.2 for i in x], added_counts, width=0.4, label="words added (not represented)")
ax.set_xticks(list(x))
ax.set_xticklabels([r.replace(" ", "\n") for r in receivers], fontsize=8)
ax.set_ylabel("words")
ax.set_title("Same representation, different reconstruction (synthetic counts)")
ax.legend()
plt.tight_layout()
plt.show()
```

## What this establishes

The chart shows reconstructed word counts and added words for each receiver.
These are synthetic demonstration counts illustrating the mechanism, not
measurements of real readers. The real point is structural: identical input
produces different output because reconstruction depends on the receiver.
