*Language: When Intelligence Enters the Channel* — Part II — Too Much Information

This notebook is a practical companion to Chapter 14. It is not an abbreviated version of the chapter; it demonstrates the chapter's mechanism with executable, offline code.

## What you will learn
- That a long source must be decomposed into source-linked information units.
- That a fixed window may not carry a relationship between distant moments.
- That decomposition should not smuggle a relevance verdict into the record.

## Why this mechanism matters
Context: video decomposition into source-linked information units.


```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## A video is not an hour: decomposition into information units

A one-hour conference talk on a consensus protocol has three key moments: the
fault definition at minute 4, the two-round-trip demonstration at minute 42, and
the unbounded-delay qualification at minute 53. No single thirty-second window
carries the relationship between them.

This notebook decomposes the talk into source-linked information units and shows
why keeping only minute 42 is misleading.


```python
fixture = FIX["video_talk"]
print("Source:", fixture["source"])
print()
print("Moments:")
for moment in fixture["moments"]:
    print(f"  Minute {moment['minute']:2d}: {moment['content']}")
```

## Experiment: linked units preserve distant evidence

Three units link the definition, demonstration and qualification. A summary that
keeps only minute 42 removes the timing qualification and can be substantively
misleading without inventing a single new number.


```python
units = fixture["units"]
print("Information units (source-linked):")
for u in units:
    print(f"  {u['unit_id']}: span={u['span']} claims={u['claims']} depends_on={u['dependencies']}")

print()
print("Misleading summary:", fixture["misleading_summary"])
print("Faithful summary:", fixture["faithful_summary"])
```

## What this establishes

A usable representation must link distant evidence while preserving the route back
to each moment. The timestamps and transcript are constructed for illustration,
not observations from an actual recording.
