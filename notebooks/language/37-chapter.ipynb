## Experiment: offline Language trace

The trace shows: expression → proposal → human correction → acceptance → transfer
to a fresh recipient → use within granted authority. The original conversation
is excluded from the transfer; only accepted state, public semantics and
evidence digests are sent. Simulated publication is BLOCKED without a separate
grant.

This uses the Language Runtime (LANG-RT-01) design concepts — deterministic mock
participants, no live inference — as the basis for the demonstration.


```python
import hashlib
import json

def state_id(state):
    return hashlib.sha256(json.dumps(state, sort_keys=True).encode()).hexdigest()[:12]

trace = [
    {"stage": "expression", "actor": "human", "state": "intention submitted"},
    {"stage": "proposal", "actor": "interpreter", "state": "interpretation proposed, bound to source"},
    {"stage": "correction", "actor": "human", "state": "intent corrected"},
    {"stage": "acceptance", "actor": "human", "state": "accepted into shared Language State"},
    {"stage": "transfer", "actor": "system", "state": "accepted state + digests sent to fresh recipient"},
    {"stage": "use", "actor": "recipient", "state": "readback proposes action; simulation BLOCKED without grant"},
]

print("Offline Language trace (deterministic, mock participants):")
for i, step in enumerate(trace, 1):
    print(f"  {i}. {step['stage']:12s} [{step['actor']}] {step['state']}")
```

## What this establishes

An intention can become an agreed semantic state traceable to source, and another
agent can use that state only within its granted authority. This is a bounded
demonstration of the minimum research claim — not proof that the full Language
architecture works in deployment.
