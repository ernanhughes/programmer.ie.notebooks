## The person changes during the day: context vs overlay

Iris asks for a five-minute briefing before a meeting. Her standing policy asks for
implementation detail, but five minutes is her stated budget for this encounter,
not a standing preference for brevity.

Durable history, current context and session overlay are distinct. A session-scoped
overlay should disappear when the session ends.


```python
layers = ["durable history", "current context", "session overlay"]
print("Three distinct layers:")
for layer in layers:
    print(f"  - {layer}")
print()
print("durable history != current context != session overlay")
```

## Experiment: temporary condition, not durable property

The channel puts failure risks first and retains each decisive qualification. If the
constraints cannot be explained safely in five minutes, it says so. Tomorrow,
when Iris asks for the full derivation, today's urgency must not govern that
request.


```python
session = {
    "standing_policy": "implementation detail and equations when they affect the decision",
    "current_context": "five minutes before an architecture review",
    "session_overlay": "failure risks first, qualifications retained, full derivation available",
}

print("Session state (constructed illustration):")
for key, value in session.items():
    print(f"  {key}: {value}")

print()
print("Overlay applies only to this session; it does not update standing policy.")
```

## What this establishes

Temporary conditions should alter communication without becoming durable properties
of the person. The distinction is architectural, not a measurement of Iris's
attention span.
