```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## Capstone II — The Personal Information Radar

Iris's afternoon includes four items: a possibly weakening paper, a library update,
a familiar survey, and an incident report affecting tonight's deployment. The
Radar decides what reaches her, at what resolution and when.

The pipeline: authorised sources → eligible candidates → standing interest/task →
attention qualification → delivery permission → reader.


```python
candidates = FIX["subscription_candidates"]
print("Authorised source candidates:")
for c in candidates:
    print(f"  {c['candidate_id']}: {c['title'][:50]}...")
    print(f"      relation={c['relation']}  relevance={c['task_relevance']}")
```

## Experiment: attention pipeline with audit envelope

The Radar selects candidates for interruption, queue, digest or withhold. Each
decision carries candidate_id, source_id, coverage, eligibility, relation,
relevance and necessity — decision dimensions plus an audit envelope, not a
single priority number.


```python
decisions = {
    "c4": "NOW (incident affecting tonight's deployment — verified blocker)",
    "c1": "QUEUE (possible weakening — inspect at first break)",
    "c2": "DIGEST (library update — not urgent)",
    "c3": "WITHHOLD (familiar material, low novelty)",
}

print("Radar delivery decisions (constructed illustration):")
for cid, decision in decisions.items():
    print(f"  {cid}: {decision}")

print()
print("Each decision carries an audit envelope: candidate, source, coverage,")
print("eligibility, relation, relevance, necessity — not a single score.")
```

## What this establishes

The Radar watches authorised sources and declared interests, not Iris herself. Any
delivery context needs a separate permission. These decisions are constructed, not
observed from a live deployment.
