```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## Capstone I — The Semantic Browser

A twenty-page incident report contains a timeline, log excerpts, five candidate
causes and a disputed conclusion. Different tasks need different representations.
This notebook assembles the Part I pieces: source, interpretation, declared
obligations, a task, a chosen representation, and a release check.


```python
fixture = FIX["incident_report"]
print("Source version:", fixture["source_version"])
print("Source:", fixture["source"])
print()
print("Candidate causes:", ", ".join(fixture["candidate_causes"]))
print("Conditional:", fixture["conditional"])
print("Missing measurement:", fixture["missing_measurement"])
```

## Experiment: source-linked record, obligations, release gate

For a manager deciding whether to approve a fix, the obligations are declared
before any view is generated. A candidate diagram that labels hypothesis three
"excluded" without the lag condition fails the release gate.

The unchanged source remains available as a fallback. Pre-display checks cannot
certify reader understanding.


```python
obligations = fixture["preservation_obligations"]
candidate_view = "Hypothesis three is excluded."  # unfaithful: drops condition

def passes_gate(view, obligations):
    """Toy gate: view must not assert an unconditional exclusion."""
    return "excluded" not in view.lower() or "only if" in view.lower()

print("Candidate view:", candidate_view)
print("Passes release gate:", passes_gate(candidate_view, obligations))
print()
print("Reason: the view drops the lag condition and the missing measurement.")
print("Gate action: refuse and return to source (fallback to unchanged report).")
```

## What this establishes

The Semantic Browser is a proposed architecture, not a completed system. The gate
checks whether declared obligations survive, but cannot certify what the reader
will understand. This notebook uses a toy gate — real checks require evidence.
