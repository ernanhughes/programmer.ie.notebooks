```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## Too much information: candidate abundance

When candidate generation becomes cheaper than inspection, the bottleneck moves
from "can the system make something to show?" to "which candidate, if any,"
deserves this person's limited attention?".

This notebook separates six distinct stages: exists, accessible with permission,
relevant to this task, selected under a budget, presented to the reader, noticed
or used.


```python
stages = ["exists", "accessible with permission", "relevant to this task",
          "selected under a budget", "presented to the reader", "noticed or used"]

print("Candidate pipeline stages:")
for i, stage in enumerate(stages, 1):
    print(f"  {i}. {stage}")
print()
print("exists != accessible != relevant != selected != presented != noticed")
```

## Experiment: six stages are distinct

An item may exist in a corpus but not be accessible to the reader's permissions. It
may be accessible but irrelevant to the task. It may be relevant but not selected
under the budget. Each stage is a separate decision with separate evidence.


```python
candidates = FIX["subscription_candidates"]
budget = 2
selected = sorted(candidates,
                  key=lambda c: {"critical": 3, "high": 2, "medium": 1, "low": 0}[c["task_relevance"]],
                  reverse=True)[:budget]

print(f"Total candidates: {len(candidates)}")
print(f"Selected under budget {budget}: {[c['candidate_id'] for c in selected]}")
print(f"Not selected: {[c['candidate_id'] for c in candidates if c not in selected]}")

for c in candidates:
    in_sel = c in selected
    print(f"  {c['candidate_id']}: relevance={c['task_relevance']:8s} selected={in_sel}")
```

## What this establishes

Selection under a budget admits some candidates and withholds others. The decision
dimensions plus audit envelope matter more than a single priority number. These
are synthetic rankings, not measured selection outcomes.
