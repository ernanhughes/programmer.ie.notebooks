## A society of proxies: local authorisation is not global safety

Five hundred researchers each authorise a proxy to look for collaborators. Each
submits twenty introduction proposals — ten thousand total. No individual grant
is violated, but the aggregate burden can overload recipients and bury valuable
connections.

Sender permission differs from receiver admission, presentation and human
attention. Pairwise-safe interactions do not imply population safety.


```python
proxies = 500
proposals_per_proxy = 20
total = proxies * proposals_per_proxy

print(f"Proxies: {proxies}")
print(f"Proposals per proxy: {proposals_per_proxy}")
print(f"Total proposals: {total}")

print()
print("Daniel appears in 400 proposals (constructed distribution).")
print("His receiving proxy can filter or queue, but capacity is still consumed.")

print()
print("local authorisation != global safety")
```

## Experiment: aggregate burden without individual violation

The question is not merely whether each proxy obeys its grant. It is whether the
channel can handle total workload, respect receivers' policies, distribute
opportunity and make failures contestable. None of these is demonstrated.


```python
properties = [
    "channel can handle total workload",
    "receivers' policies respected",
    "opportunity distributed acceptably",
    "failures contestable",
]

print("Properties not yet demonstrated for the Language channel:")
for p in properties:
    print(f"  - {p}")
```

## What this establishes

An ecosystem is not a society. Locally authorised requests can create
population-level effects that no individual grant controls. This chapter separates
the unit of analysis from sender permission to receiver and channel policy.
