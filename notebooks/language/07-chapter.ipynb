## Meaning becomes the source: source vs semantic record vs rendering

A payments retry policy is rendered three times in one week: a state diagram on
Monday, a timeline on Wednesday, a table on Friday. Each run starts from the
same prose but with no shared interpretation. The views can disagree.

This notebook introduces three objects: **source** (the canonical version),
**semantic record** (a derived, fallible, source-linked account), and
**rendering** (a task-specific presentation from the record).


```python
source_version = "payments-retry-policy-v1"
source_passage = "Retry: at most 3 attempts for non-idempotent requests. Timeout handling unresolved."

semantic_record = [
    {"id": "claim-1", "text": "max 3 attempts for non-idempotent requests", "source_span": "passage", "uncertainty": None},
    {"id": "claim-2", "text": "timeout handling unresolved", "source_span": "passage", "uncertainty": "missing measurement"},
]

print("Source version:", source_version)
print("Source passage:", source_passage)
print()
print("Semantic record entries:")
for entry in semantic_record:
    print(f"  {entry['id']}: {entry['text']} (uncertainty: {entry['uncertainty']})")
```

## Experiment: drift without a shared record

Direct generation creates three views from three separate readings:
- Monday: diagram shows partition exception → dead-letter queue.
- Wednesday: timeline calls third retry "final" (omits partition).
- Friday: table omits the exception row for space.

Agreement between views would not establish correctness — all three could repeat
the same mistaken reading. A shared record lets all views be checked against one
account.


```python
views = [
    {"day": "Monday", "rendering": "state diagram", "shows_partition_exception": True, "calls_third_final": False},
    {"day": "Wednesday", "rendering": "timeline", "shows_partition_exception": False, "calls_third_final": True},
    {"day": "Friday", "rendering": "table", "shows_partition_exception": False, "calls_third_final": False},
]

print("View comparisons (synthetic illustration):")
for v in views:
    print(f"  {v['day']:8s} {v['rendering']:12s} partition={v['shows_partition_exception']} final={v['calls_third_final']}")

all_agree = all(v["shows_partition_exception"] == views[0]["shows_partition_exception"] for v in views)
print(f"\nAll views agree on partition exception: {all_agree}")
print("Agreement does not establish correctness against the source.")
```

## What this establishes

Without a shared semantic record, views can silently disagree. With a shared record,
each rendering can cite the record entry it came from. Traceability is required
but is not proof of correctness — this is the chapter's central distinction.
