## What the system believes about you must be visible

Iris opens a talk about a consensus algorithm she has not used. The channel folds
the failure-assumptions section, saying consensus foundations are treated as
already known — inferred from earlier architecture reviews.

A belief that Iris knows consensus is not the same as a decision to fold this
explanation. The stored belief, the decision that used it and the effect must be
distinguished.


```python
belief = {
    "content": "Iris knows consensus foundations",
    "origin": "inferred from earlier architecture reviews",
    "supported_within_scope": True,
    "usable_here": False,  # does not establish familiarity with this algorithm
}

decision = {
    "action": "fold the failure-assumptions section",
    "used_belief": "belief content",
    "effect": "Iris does not see the scope qualification",
}

print("Stored belief:")
for key, value in belief.items():
    print(f"  {key}: {value}")
print()
print("Decision that used the belief:")
for key, value in decision.items():
    print(f"  {key}: {value}")
```

## Experiment: visible is not authorised

The explanation is plausible — that is the danger. Plausibility is not a record of
which belief actually caused the omission. The system must show where each belief
came from and what it changes.


```python
distinctions = ["stored belief", "decision that used it", "actual effect"]
print("Three objects to keep separate:")
for d in distinctions:
    print(f"  - {d}")

print()
print("Visible != authorised: showing the belief does not grant authority to use it.")
```

## What this establishes

Inferred beliefs must show origin, scope and what they change. Visible is not the
same as authorised. These cases are constructed, not observed measurements.
