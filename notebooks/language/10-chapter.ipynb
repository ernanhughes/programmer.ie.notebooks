## What you will learn
- That transformation produces survival, loss, addition and gain.
- That gain requires reader or task evidence, not just output inspection.
- That a faithful preview must preserve conditions and qualifications.

## Why this mechanism matters
Conceptual: survival / loss / addition / gain audit on a policy transformation.


```python
import json
from pathlib import Path
import sys

# Resolve the repository root by walking parents from the current working directory.
# This mirrors the Retrieval book notebook convention: _support walks up looking for
# content/books/language/_index.md, so no fixed working directory is required.
ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "content" / "books" / "language" / "_index.md").is_file():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError("repository root not found")

# Make this notebook's shared helper importable.
NOTEBOOK_DIR = ROOT / "notebooks" / "books" / "language"
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import _support
ROOT = _support.repo_root()
print("repo root:", ROOT)

# Load deterministic fixtures (constructed / synthetic; no real user data).
FIXTURES = ROOT / "notebooks" / "books" / "language" / "fixtures" / "language-fixtures.json"
with open(FIXTURES, encoding="utf-8") as f:
    FIX = json.load(f)
print("fixture_id:", FIX["fixture_id"])
print("seed:", FIX["seed"])
```

## What survived the transformation? Survival / loss / addition / gain

A system summarizes a payments retry policy: "Failed payments are retried up to
three times." The source says at most three attempts total for non-idempotent
requests, timeout handling unresolved. The summary loses the condition, the
universality limit, and the unresolved timeout.

This notebook audits a transformation using four outcomes: survival, loss,
addition and gain.


```python
fixture = FIX["incident_report"]
print("Source:", fixture["source"])
print()
print("Unfaithful summary:", fixture["unfaithful_summary"])
print("Faithful preview:", fixture["faithful_preview"])
print()
print("Preservation obligations:")
for obligation in fixture["preservation_obligations"]:
    print(f"  - {obligation}")
```

## Experiment: survival / loss / addition / gain audit

The unfaithful summary deletes the condition (hypothesis three excluded only if lag
below threshold) and the missing measurement. The faithful preview preserves
them. Gain (easier to read) requires reader or task evidence — it cannot be
established from the output alone.


```python
audit = {
    "survival": "conditional status preserved in faithful preview; lost in unfaithful summary",
    "loss": "replication-lag measurement missingness removed in unfaithful summary",
    "addition": "unfaithful summary adds an unconditional exclusion not in source",
    "gain": "unmeasured — requires reader/task evidence, not output inspection",
}
for kind, finding in audit.items():
    print(f"{kind.upper():8s}: {finding}")
```

## What this establishes

A fluent output can fail the preservation check while passing a fluency check. The
audit method declares obligations before generation and reports outcomes after.
Gain is not addition — gain needs evidence of reader benefit.
