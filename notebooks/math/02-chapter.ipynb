# Refute Fermat's prime formula with one division

*Adversarial Mathematics From First Principles*, [chapter 2](https://programmer.ie/books/math/02-chapter/) · [worked example](https://programmer.ie/examples/math/02-chapter/)

Run the cells in order. Everything below the program is taken from the laboratory's recorded, hash-verified run; the program is a small, self-contained demonstration of the same mathematics.

Fermat believed every number of the form 2^(2^n) + 1 is prime. The first five
are, and he stopped there. The chapter's claim is that a universal statement
needs only one counterexample, and that five confirming cases add nothing
beyond themselves. This program finds the counterexample Euler reported in
1732: the sixth number is divisible by 641.

## Run it

Standard library only; nothing to install.

````python
"""Fermat's numbers and Euler's 641, in exact integer arithmetic (standard library only)."""


def fermat(n):
    return 2 ** (2 ** n) + 1


def smallest_factor(m):
    """The smallest divisor of m greater than 1, by trial division."""
    d = 2
    while d * d <= m:
        if m % d == 0:
            return d
        d += 1
    return m


print("n  2^(2^n) + 1   smallest factor > 1")
for n in range(6):
    f = fermat(n)
    p = smallest_factor(f)
    print(f"{n}  {f:<12} {p:<10} {'prime' if p == f else 'composite'}")

f5 = fermat(5)
print()
print(f"2^32 + 1            = {f5}")
print(f"remainder mod 641   = {f5 % 641}")
print(f"cofactor            = {f5 // 641}")
print(f"641 * 6700417 == F5 : {641 * 6700417 == f5}")
print(f"6700417 is prime    : {smallest_factor(6700417) == 6700417}")
print()
# Failure control: a wrong divisor must not pass.
print(f"control: remainder mod 643 = {f5 % 643} (643 does not divide F5)")
````

## What the output shows

Five primes, then a composite. The remainder 0 is the whole argument: since
1 < 641 < 4294967297, the sixth number has a divisor other than 1 and itself,
so the claim "every such number is prime" is false. The control matters as much
as the result: a check that would also accept 643 would prove nothing.
