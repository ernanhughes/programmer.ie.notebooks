# Read what a Lean statement actually says

*Adversarial Mathematics From First Principles*, [chapter 15](https://programmer.ie/books/math/15-chapter/) · [worked example](https://programmer.ie/examples/math/15-chapter/)

Run the cells in order. The program is a small, self-contained demonstration of the chapter's mathematics; the output saved with it is what it printed when this notebook was built.

A formal statement's symbols mean what the library says they mean, and
those are not always what a reader brings. This core-Lean file shows four
places where that matters: natural-number subtraction, division by zero, a
hypothesis that can never hold, and a theorem whose name says one thing and
whose statement says another.

```python
%%writefile formal_propositions.lean
-- What a formal proposition says: the symbols mean what the library says (core Lean 4).

-- Natural-number subtraction is truncated: there is no -1 in Nat.
#eval (2 - 3 : Nat)            -- 0
#eval (2 - 3 : Int)            -- -1

-- Division by zero is defined, and is 0. A total function, not a fact about numbers.
#eval (7 / 0 : Nat)            -- 0

-- So an identity over the integers can fail over the naturals.
example : ∀ m n : Int, (m - n) + n = m := by intro m n; omega        -- true
example : ¬ ∀ m n : Nat, (m - n) + n = m := by                         -- false on Nat
  intro h
  have := h 2 3
  omega

-- A vacuous hypothesis: the kernel accepts this, correctly. It says nothing.
theorem anything_from_false (h : 0 = 1) : 2 + 2 = 5 := by omega

-- The name says nothing either. This "fermat_is_false" proves something else entirely.
theorem fermat_is_false : 1 + 1 = 2 := rfl
#check @fermat_is_false        -- the statement is what counts, not the name
```

```python
!lean-4.35.0-linux/bin/lean formal_propositions.lean
```

## What the output shows

On `Nat`, 2 − 3 is 0 and 7 / 0 is 0: total functions, not facts about numbers.
So `(m − n) + n = m` is an identity on `Int` and false on `Nat`, and Lean
proves both. The theorem with hypothesis `0 = 1` is accepted, correctly, and
says nothing. `fermat_is_false` proves `1 + 1 = 2`: only `#check`, the
statement, tells you what was proved.

## Not shown

- Core Lean only; Mathlib's real-number conventions (√ of a negative is 0) are described in the chapter, not shown here.
- Lean accepting a file shows the statements are proved, not that they are the statements anyone wanted.
