# Take two limits in both orders and get two answers

*Adversarial Mathematics From First Principles*, [Part II, chapter 3](https://programmer.ie/books/math/12-chapter/) · [worked example](https://programmer.ie/examples/math/12-chapter/)

Run the cells in order. The program is a small, self-contained demonstration of the chapter's mathematics; the output saved with it is what it printed when this notebook was built.

The chapter's smallest example is xⁿ on [0, 1]: taking the limit in n first
and then x → 1 gives 0; the other order gives 1. This program shows the two
orders, measures how slowly the convergence happens near x = 1, and sums Abel's
series near its jump.

## Run it

Standard library only; nothing to install.

```python
"""Limits that do not commute: x^n on [0, 1], and Abel's series (standard library only)."""
import math

print("x^n: the two orders of limit")
print(f"  limit over n first, then x -> 1:  x^1000 at x = 0.999 is {0.999 ** 1000:.2e}, and -> 0 as n grows")
print(f"  limit over x first, then n:       x^n at x = 1 is {1.0 ** 1000} for every n")

print("\nPointwise, not uniform: the n needed for x^n < 1/2 at x = 1 - 1/k")
for k in (10, 100, 1000, 10000):
    x = 1 - 1 / k
    n = math.ceil(math.log(0.5) / math.log(x))
    print(f"  k = {k:>5}: n = {n:>5}   (k ln 2 = {k * math.log(2):8.1f})")
print("  sup over [0,1) of |x^n - 0| is 1 for every n: no single n serves every x.")


def abel(x, terms):
    """Partial sum of sum_{n>=1} (-1)^(n+1) sin(n x) / n, which converges to x/2 on (-pi, pi)."""
    return math.fsum((-1) ** (n + 1) * math.sin(n * x) / n for n in range(1, terms + 1))


print("\nAbel's series: each term continuous, the sum x/2 on (-pi, pi), but 0 at x = pi")
for x in (3.0, 3.1, 3.14, math.pi):
    print(f"  x = {x:.5f}: partial sum (100000 terms) {abel(x, 100000): .5f}   x/2 = {x / 2:.5f}")
print("  The limit jumps from pi/2 to 0 at x = pi: a convergent sum of continuous functions, discontinuous.")
```

## What the output shows

Near x = 1 the n needed for xⁿ < 1/2 grows like k ln 2 at x = 1 − 1/k, without
bound: the convergence is pointwise, not uniform, and that is exactly the
hypothesis Cauchy's theorem needs. Abel's series sums to x/2 just below π and
to 0 at π: every term is continuous and the series converges everywhere, yet
the sum jumps.

## Not shown

- Partial sums at finitely many points illustrate the limit; they do not prove convergence or its value. Part I, chapter 2 has Lean check the finite identity behind Abel's series.
- Floating-point sums of 100,000 terms carry rounding error; the jump is far larger than that error.
