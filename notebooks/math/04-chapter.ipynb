# Find where Haraux's two collision times start

*Adversarial Mathematics From First Principles*, [chapter 4](https://programmer.ie/books/math/04-chapter/) · [worked example](https://programmer.ie/examples/math/04-chapter/)

Run the cells in order. Everything below the program is taken from the laboratory's recorded, hash-verified run; the program is a small, self-contained demonstration of the same mathematics.

A 2024 paper by A. Haraux derives two collision times for one scenario,
D/(c + v) and D/c, and argues that their difference is a contradiction in the
mathematics of special relativity. Two numbers that differ contradict each
other only if they answer the same question. This program computes the events
exactly, at v = 3/5 c, and asks where each time starts.

## Run it

Standard library only; nothing to install.

````python
"""Haraux's two collision times, exactly (standard library only).

Units c = 1, D = 1. The photon is emitted at x = 0 at S-time 0 (event E), towards
M, which is at x = D at S-time 0 (event A) and moves at -v. They meet at event B.
M's rest frame S' is reached by the Lorentz transformation for velocity -v.
"""
from fractions import Fraction as F

v = F(3, 5)
gamma = F(5, 4)                      # 1 / sqrt(1 - v^2), exact for v = 3/5
assert gamma * gamma * (1 - v * v) == 1


def to_m_frame(t, x):                # S -> S'
    return gamma * (t + v * x), gamma * (x + v * t)


E = (F(0), F(0))
A = (F(0), F(1))
B = (1 / (1 + v), 1 / (1 + v))       # the photon reaches M at S-time D / (c + v)
for name, ev in (("E: photon emitted", E), ("A: M at S-time 0", A), ("B: photon meets M", B)):
    print(f"{name:20} S {tuple(map(str, ev))!s:14} S' {tuple(map(str, to_m_frame(*ev)))}")

print()
print(f"E and A simultaneous in M's frame? {to_m_frame(*E)[0] == to_m_frame(*A)[0]}")
x_m = gamma * 1                      # M rests at x' = gamma * D in S'
print(f"photon speed in M's frame: {x_m / to_m_frame(*B)[0]}")

t1 = B[0] - E[0]                     # T1 = D/(c + v): from E to B
s_star = (gamma - 1 / gamma)         # the S'-slice on which the photon is alpha*D from M
t_start = gamma * (s_star - v * x_m) # S-time of M's event on that slice
t2 = B[0] - t_start                  # T2: from that event to B
print()
print(f"T1 = D/(c+v) = {t1}, starting at E (S-time 0)")
print(f"T2 = {t2} = D/c? {t2 == 1}, starting at M's event at S-time {t_start}")
print(f"the two start events differ by {t2 - t1} in S-time; T2 - T1 = {t2 - t1}")
print()
# Failure control: the paper's final step says T1 and T2 time the same interval.
print(f"control: same start event? {t_start == E[0]} -> the claim is refused")
````

## What the output shows

Both times end at the same event, B, but they start at different ones: T1 at
the emission E, T2 at an event on M's worldline 3/8 earlier. Their difference
is exactly the gap between the start events, so it is not a contradiction. The
question is whether the derivation exposes an inconsistency in the
mathematics, not whether relativity describes the world.
