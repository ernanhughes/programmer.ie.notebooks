# Refute a 'for all' claim with one exact witness

*Adversarial Mathematics From First Principles*, [Part II, chapter 1](https://programmer.ie/books/math/10-chapter/) · [worked example](https://programmer.ie/examples/math/10-chapter/)

Run the cells in order. The program is a small, self-contained demonstration of the chapter's mathematics; the output saved with it is what it printed when this notebook was built.

The chapter's claim is that a universal statement is refuted by one valid
witness, and that passing tests are weak evidence because the inputs people
try first are often degenerate. The claim under test is that the shear
t′ = t, x′ = x + vt preserves t² − x². This program tries the convenient
inputs, then a grid, then the residual itself, all in exact arithmetic.

## Run it

Standard library only; nothing to install.

```python
"""Every, some, none: why passing tests are weak evidence (standard library only).

Claim: the shear t' = t, x' = x + v t preserves t^2 - x^2 for all t, x, v.
Exact arithmetic with fractions, so a zero means zero.
"""
from fractions import Fraction as F


def residual(t, x, v):
    """(t'^2 - x'^2) - (t^2 - x^2) for the shear."""
    return (t * t - (x + v * t) ** 2) - (t * t - x * x)


print("Convenient inputs (each one degenerate):")
for t, x, v, why in [(0, 5, F(1, 2), "t = 0"), (3, 7, 0, "v = 0"), (2, F(-1, 2), F(1, 2), "x = -vt/2")]:
    print(f"  t={t!s:>4} x={x!s:>5} v={v!s:>4}  residual {residual(F(t), F(x), F(v))}   ({why})")

print("\nA grid of ordinary inputs, t, x in -3..3 and v in {1/4, 1/2, 3/4}:")
grid = [(F(t), F(x), v) for t in range(-3, 4) for x in range(-3, 4) for v in (F(1, 4), F(1, 2), F(3, 4))]
fails = [p for p in grid if residual(*p) != 0]
print(f"  {len(grid) - len(fails)} of {len(grid)} pass; {len(fails)} fail")

print("\nThe residual is -v t (2x + v t): it vanishes only on a thin set.")
assert all(residual(t, x, v) == -v * t * (2 * x + v * t) for t, x, v in grid)
print("  identity holds at every grid point: True")

w = (F(1), F(0), F(1, 2))
print(f"\nWitness t=1, x=0, v=1/2: residual {residual(*w)}. One point refutes 'for all'.")

print("\nPythagoras with the right angle deleted, A=(0,0) B=(1,0) C=(1,1):")
d2 = lambda p, q: (p[0] - q[0]) ** 2 + (p[1] - q[1]) ** 2
A, B, C = (0, 0), (1, 0), (1, 1)
print(f"  d2(B,C) = {d2(B, C)}, d2(A,B) + d2(A,C) = {d2(A, B) + d2(A, C)}, residual {d2(B, C) - d2(A, B) - d2(A, C)}")

print("\nOrder of quantifiers, over the integers:")
print("  forall x exists y, y > x: for each x, y = x + 1 works, e.g.",
      ", ".join(f"x={x}: y={x + 1}" for x in (-5, 0, 7)))
print("  exists y forall x, y > x: every candidate y fails at x = y, e.g.",
      ", ".join(f"y={y}: x={y} gives {y} > {y} False" for y in (0, 10**6)))
```

## What the output shows

The three convenient inputs all pass: each is a degenerate case (t = 0, v = 0,
or x = −vt/2). On an ordinary grid most points fail, and the residual explains
why: it is −vt(2x + vt), zero only on a thin set. One witness, t = 1, x = 0,
v = 1/2, with residual −1/4, refutes the "for all". The same pattern refutes
Pythagoras with the right angle deleted. The last lines show why the order of
"for all" and "there exists" matters.

## Not shown

- The grid is a test, not a proof: the residual identity is checked at 147 points here; in general it is a polynomial identity to be proved algebraically.
- The witness refutes only the claim as stated. It says nothing about which repair is right (that is the next chapter).
