# Settle Gabriel's paint: one slogan, two kinds of paint

*Adversarial Mathematics From First Principles*, [Part II, chapter 10](https://programmer.ie/books/math/19-chapter/) · [worked example](https://programmer.ie/examples/math/19-chapter/)

Run the cells in order. The program is a small, self-contained demonstration of the chapter's mathematics; the output saved with it is what it printed when this notebook was built.

The chapter separates three questions: is the mathematics true under its
assumptions, does it describe what we meant, and what supports that
correspondence? "You can fill Gabriel's horn but not paint it" mixes them. This
program puts numbers on each part of the slogan.

## Run it

Standard library only; nothing to install.

```python
"""Mathematics and the world: Gabriel's horn and the paint (standard library only).

Horn: radius 1/x for x >= 1, rotated about the x-axis. Lengths in metres.
"""
import math

MOLECULE = 1e-9   # a generous size for a paint molecule: 1 nanometre

print("1. True under its assumptions: the two integrals (exact forms)")
for b in (10, 1e3, 1e6, 1e9):
    print(f"  up to x = {b:<8.0e} volume pi(1 - 1/b) = {math.pi * (1 - 1 / b):.9f}   area > 2 pi ln b = {2 * math.pi * math.log(b):8.2f}")

print("\n2. Painting the outside with a coat of fixed thickness d needs at least d * area:")
for d in (1e-3, 1e-6):
    for b in (1e3, 1e9, 1e100):
        print(f"  d = {d:g} m, up to x = {b:<6.0e}: paint > {d * 2 * math.pi * math.log(b):.3e} m^3")
print("  As b grows without bound, so does the paint: any fixed thickness needs infinite paint.")

print("\n3. 'Filling it' coats the inside with paint whose thickness is the radius 1/x:")
x_mol = 1 / MOLECULE
print(f"  the tube is narrower than a molecule beyond x = {x_mol:.0e} m")
print(f"  real paint fills at most the volume up to there: {math.pi * (1 - 1 / x_mol):.9f} m^3 (< pi)")
print("  and cannot reach the rest of the horn, which is infinitely long.")

print("\n4. Mathematical paint: a coat of thickness 1/x over the whole horn has volume pi, finite;")
print("   a coat of any fixed thickness has infinite volume. No contradiction: the slogan's")
print("   'paint' meant two different things.")
```

## What the output shows

The volume integral settles on π and the area grows without bound: that is
the mathematics. A coat of any fixed thickness needs at least thickness ×
area, which has no bound. "Filling" coats the inside with paint whose
thickness is the radius 1/x, which shrinks below a molecule at x = 10⁹ m.
With mathematical paint both halves hold; with real paint both fail. The
paradox is in the slogan, not the integrals.

## Not shown

- The integrals are taken in their exact forms; Part I, chapter 8 has Lean prove the convergence and divergence.
